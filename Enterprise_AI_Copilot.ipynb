# 1️⃣ Zero-Shot Prompting

## What is Zero-Shot Prompting?

Zero-Shot Prompting is a technique where the AI model is asked to perform a task without providing any examples. The model relies entirely on its pre-trained knowledge to understand the instruction and generate the response.

In this section, we evaluate how Gemini performs a task using only a single instruction.

````python
# Zero-Shot Prompting Example

prompt = """
Write a professional email to the HR department requesting one day of leave due to a medical appointment.
"""

try:
    response = client.models.generate_content(
        model="models/gemini-3.5-flash",
        contents=prompt
    )
    print(response.text)

except Exception as e:
    print("Primary model busy. Trying backup model...\n")

    response = client.models.generate_content(
        model="models/gemini-2.0-flash",
        contents=prompt
    )

    print(response.text)
````

### Analysis

The Gemini model successfully generated a professional leave request email using only a single instruction without any examples. This demonstrates the effectiveness of Zero-Shot Prompting, where the model relies on its pre-trained knowledge to understand the task and produce a well-structured response.

# 2️⃣ Few-Shot Prompting

## What is Few-Shot Prompting?

Few-Shot Prompting is a technique where the AI model is provided with a few examples before asking it to perform a new task. These examples help the model understand the expected writing style, format, and response pattern.

In this section, we provide sample emails before requesting the AI to generate a new professional email.

````python
# Few-Shot Prompting Example

prompt = """
Example 1

Request:
Write a leave email for one day due to fever.

Response:
Subject: Leave Request

Dear HR,

I am feeling unwell due to fever and would like to request one day leave. I will resume work tomorrow.

Thank you.

Best Regards,
Ali


----------------------------

Example 2

Request:
Write a leave email for a family event.

Response:
Subject: Leave Request

Dear HR,

I would like to request one day leave to attend an important family event. I have completed my pending work.

Thank you.

Best Regards,
Sara


----------------------------

Now generate a professional leave email for a medical appointment.
"""

response = client.models.generate_content(
    model="models/gemini-3.5-flash",
    contents=prompt
)

print(response.text)
````

### Analysis

Unlike Zero-Shot Prompting, the model was provided with multiple examples before receiving the final task. The generated response followed the writing style and structure demonstrated in the examples, producing a more consistent and professional output. This shows how Few-Shot Prompting improves response quality by guiding the model with sample patterns.

# 3️⃣ Chain of Thought (CoT) Prompting

## What is Chain of Thought Prompting?

Chain of Thought (CoT) Prompting encourages the AI model to explain its reasoning step by step before providing the final answer. Instead of directly producing a response, the model breaks down the problem into logical steps, improving reasoning accuracy for complex tasks.

In this section, the AI is instructed to think step by step before generating the final response.

````python
# Chain of Thought Prompting Example

prompt = """
A company wants to organize an AI Workshop.

Think step by step before answering.

Include:

1. Objectives
2. Required Resources
3. Expected Participants
4. Budget Considerations
5. Final Recommendation

Provide the final answer in a professional report format.
"""

response = client.models.generate_content(
    model="models/gemini-3.5-flash",
    contents=prompt
)

print(response.text)
````

### Analysis

In this example, the AI model was instructed to think step by step before generating the final response. Instead of directly producing an answer, it organized its reasoning into logical stages and then generated a structured report. This demonstrates how Chain of Thought Prompting improves reasoning quality and produces more detailed, organized, and reliable outputs for complex tasks.

# 4️⃣ Tree of Thoughts (ToT)

## What is Tree of Thoughts?

Tree of Thoughts (ToT) is an advanced prompting technique where the AI explores multiple possible solutions before selecting the best one. Instead of following only one reasoning path, the model considers different approaches, evaluates them, and then provides the most suitable solution.

This technique is especially useful for planning, decision-making, brainstorming, and solving complex problems.

````python
# Tree of Thoughts Prompting Example

prompt = """
A software company wants to improve employee productivity.

Think of THREE different strategies.

For each strategy:

- Explain the idea.
- Mention its advantages.
- Mention its disadvantages.

Finally, compare all three strategies and recommend the BEST one with a clear justification.
"""

response = client.models.generate_content(
    model="models/gemini-3.5-flash",
    contents=prompt
)

print(response.text)
````

### Analysis

In this example, the AI explored multiple possible solutions before selecting the most appropriate one. Instead of relying on a single reasoning path, it compared different strategies, evaluated their strengths and weaknesses, and provided a justified recommendation. This demonstrates how Tree of Thoughts Prompting supports better decision-making for complex real-world problems.

# 5️⃣ ReAct Prompting

## What is ReAct Prompting?

ReAct (Reason + Act) Prompting combines reasoning with actions. The AI first analyzes the problem, explains its reasoning process, and then decides what action should be taken to solve the problem.

This technique is widely used in modern AI agents because it enables models to think logically before performing tasks or calling external tools.

````python
# ReAct Prompting Example

prompt = """
You are an AI assistant.

Scenario:

A customer reports that they cannot log into their company account.

Your task:

Step 1: Explain your reasoning.

Step 2: Decide the best action.

Step 3: Generate the final response that should be sent to the customer.

Clearly separate Reasoning, Action, and Final Response.
"""

response = client.models.generate_content(
    model="models/gemini-3.5-flash",
    contents=prompt
)

print(response.text)
````

### Analysis

In this example, the AI first analyzed the user's problem before deciding on an appropriate action. It then generated a professional response based on its reasoning. This demonstrates the ReAct Prompting technique, where reasoning and action are combined to solve real-world tasks more effectively.

# 6️⃣ Function Calling

## What is Function Calling?

Function Calling allows an AI model to decide when a specific function should be executed instead of generating only text. The AI identifies the user's request, selects the appropriate function, and passes the required parameters.

This capability enables AI assistants to perform real-world tasks such as booking meetings, sending emails, generating reports, and querying databases.

````python
# Function Calling Simulation

def send_email(recipient, subject):
    return f"""
Function Executed Successfully

Recipient : {recipient}

Subject : {subject}

Status : Email Sent Successfully
"""

prompt = """
User Request:

Send an email to HR with the subject:
'Leave Request for Medical Appointment'

Instead of writing the email,
identify which function should be called.
"""

response = client.models.generate_content(
    model="models/gemini-3.5-flash",
    contents=prompt
)

print("AI Decision:\n")
print(response.text)

print("\n")

print(send_email(
    "hr@company.com",
    "Leave Request for Medical Appointment"
))
````

### Analysis

The AI identified that the user's request required a function instead of a normal text response. After understanding the user's intent, the appropriate email function was executed with the required parameters. This demonstrates the basic concept of Function Calling used in modern AI assistants.

# 8️⃣ Tool Calling

## What is Tool Calling?

Tool Calling enables an AI model to interact with external tools and services instead of only generating text. Depending on the user's request, the AI can choose the appropriate tool, execute it, and return the result.

Examples include searching the internet, querying databases, sending emails, scheduling meetings, and generating reports.

````python
# Tool Calling Simulation

def internet_search(query):
    return f"Searching the internet for: {query}"

prompt = """
User wants to know the latest AI trends.

Instead of answering directly,

identify the correct tool to use.
"""

response = client.models.generate_content(
    model="models/gemini-3.5-flash",
    contents=prompt
)

print("AI Decision:\n")
print(response.text)

print("\nTool Output:\n")
print(internet_search("Latest AI Trends"))
````

### Analysis

The AI recognized that the user's request required an external tool rather than generating information from its own knowledge. It selected the appropriate tool and simulated an internet search. This demonstrates the concept of Tool Calling used in modern AI assistants.

# 🔟 JSON Output Parsing

## What is JSON Output Parsing?

Many AI systems return information in JSON format because it is structured and easy for applications to process. JSON parsing converts AI-generated JSON into Python objects that can be used in programs.

````python
import json

prompt = """
Generate JSON for a student.

Name: Asia Noor

Course: AI Internship

Week: 2

Return only valid JSON.
"""

response = client.models.generate_content(
    model="models/gemini-3.5-flash",
    contents=prompt
)

print("Generated JSON:\n")
print(response.text)

print("\nParsed JSON:\n")

try:
    json_text = response.text.replace("```json", "").replace("```", "").strip()
    data = json.loads(json_text)
    print(data)
except Exception as e:
    print("Parsing Error:", e)
````

### Analysis

The AI generated structured JSON data which was successfully parsed into a Python dictionary. JSON parsing is essential for integrating AI systems with databases, APIs, and enterprise applications.

# 📧 Enterprise Email Drafting Module

## Overview

This module is designed to generate professional business emails using Google's Gemini AI.

Unlike a basic email generator, this enterprise version collects complete user information and dynamically creates a polished corporate email suitable for real-world business communication.

### User Inputs

- Sender Name
- Job Title
- Company Name
- Recipient Name
- Recipient Position
- Purpose of Email
- Tone
- Email Length

### Output

A complete business email including:

- Professional Subject
- Greeting
- Email Body
- Closing
- Signature

````python
# ==========================================
# 📧 Enterprise Email Drafting Module
# ==========================================

def draft_email():

    print("\n" + "="*70)
    print("📧 Enterprise Email Drafting Assistant")
    print("="*70)

    sender_name = input("👤 Your Name: ")

    job_title = input("💼 Job Title: ")

    company = input("🏢 Company Name: ")

    recipient_name = input("👤 Recipient Name: ")

    recipient_position = input("💼 Recipient Position: ")

    purpose = input("📝 Purpose of Email: ")

    tone = input("🎯 Tone (Formal / Professional / Friendly): ")

    length = input("📏 Length (Short / Medium / Long): ")

    prompt = f"""
You are a professional corporate email writer.

Generate a {tone} business email.

Company:
{company}

Sender Name:
{sender_name}

Job Title:
{job_title}

Recipient Name:
{recipient_name}

Recipient Position:
{recipient_position}

Purpose:
{purpose}

Email Length:
{length}

Instructions:

• Create a professional subject line.
• Write a corporate greeting.
• Write a polished email body.
• End with a professional closing.
• Include sender name and job title.
• Do NOT use placeholders like [Your Name].
• Use the provided information only.
"""

    response = client.models.generate_content(
        model="models/gemini-3.5-flash",
        contents=prompt
    )

    print("\n")
    print("="*70)
    print("📨 GENERATED EMAIL")
    print("="*70)

    print(response.text)


# Run Module
draft_email()
````

### Analysis

This enterprise email drafting module dynamically collects business information from the user and generates a polished corporate email using Gemini AI. Unlike the basic implementation, this version avoids placeholders and produces a complete, professional email that is ready for real-world business communication. The module demonstrates the practical use of Dynamic Prompt Templates, Prompt Engineering, and Large Language Models in enterprise environments.
